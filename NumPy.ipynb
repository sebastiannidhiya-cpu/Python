#Indexing and slicing

```python
a=np.array([1,2,3,4,5])
a
```

```python
print(a[3])
print(a[1:4])
print(a[-1])
print(a[:4])
print(a[2:])
```

```python
a=np.array([[1,2,3,4],[2,3,4,5]])
print(a[0])
print(a[0,1])
print(a[0][1])
```

```python
b=np.array([[1,2,3,4],[2,3,4,5],[6,7,8,9]])
print(b[1,0])
print(b[0:4,3])
```
